```python
temperature = pd.Series(
    [-2.4, 0.4, 5.7, 12.5, 18.2, 23.0,
     25.8, 26.1, 21.2, 14.8, 7.2, 0.1],
)

temperature
```

```python
temperature.index = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
                    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]        #인덱스 영문으로 변경

temperature
```

```python
temperature.size
temperature.shape
temperature.dtype
temperature.info()
```

```python
temperature.iloc[::2]
#temperature.loc[:]
temperature.iloc[::-1]
```

```python
temperature[temperature<0]
temperature[temperature>20]
t2=temperature + 10
t2
```

```python
t3= temperature.iloc[0:3] - 2
t3
```

```python
print("개수: ", temperature.count())
print("합: ", temperature.sum())
print("최대값: ", temperature.max())
print("최소값: ", temperature.min())

print("평균: ", temperature.mean())
print("중앙값: ", temperature.median())
# 12개 중복된 값이 없어서 12개 공동1위로 모두 나옴
print("최빈값: ", temperature.mode()) # 범주형 타입의 대표값으로 사용

print("분산: ", temperature.var()) # 각 데이터에서 평균을 뺀 후 제곱 → 모두 더함 → 데이터 개수로 나눔
print("표준편차: ", temperature.std()) # 분산에 루트
```

```python
gender = pd.Series(
    ["남", "여", "여", "남", "남", "여", "여", "남", "여"],
    dtype="category"
)

#print(gender)
print(gender.dtype)
print(gender.cat.categories) # 범주의 종류 확인 -> "남", "여" 확인 
print(gender.nunique()) # 범주의 개수
print(gender.value_counts()) # 범주별 데이터 개수
print("최빈값: ", gender.mode()) # 최빈값 : 범주형의 대표값은 최빈값. 평균, 중앙값등은 의미가 없다
print(gender.cat.codes) # 판다스가 내부적으로 관리하는 범주의 인코딩 값

print(gender.describe())

# 범주 추가
gender = gender.cat.add_categories(["기타"])

print(gender.cat.categories)
print(gender)
```
