```python
import pandas as pd
```

```python
#{레이블:[], 레이블:[]}

df1 = pd.DataFrame({
    "이름":["민수","지영"],
    "점수":[90,85]
})

display(df1)
```

```python
df2= pd.DataFrame([
    ["민수",90],
    ["지영",85]
    ],index=["첫번째","두번째"],columns=["이름","점수"])

display(df2)
```

```python
df3= pd.DataFrame([
    {"이름":"민수","점수":90},
    {"이름":"지영","점수":85}
])

display(df3)
```

```python
import numpy as np

df3 = pd.DataFrame(np.array([
    [1,2],
    [3,4]
]))

display(df3)
```
