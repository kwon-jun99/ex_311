```python
#문제 1. []로 열 선택
#mpg 데이터프레임에서 mpg 열만 선택하여 출력하시오.

mpg["mpg"]
```

```python
#문제 2. []로 여러 열 선택
#mpg 데이터프레임에서 mpg, weight, origin 세 개의 열을 선택하여 출력하시오.

mpg[0:7:3]
```

```python
#​문제 3. iloc로 특정 행 선택
#mpg에서 10번째 행의 데이터를 출력하시오.단, 숫자 위치를 이용하는 iloc을 사용하시오.
mpg.iloc[9]
```

```python
#문제 4. iloc로 행 + 열 선택
#mpg에서 3번째 행의 1번째 열 값을 출력하시오.

mpg.iloc[2,0]
```

```python
#문제 5. iloc로 행 + 열 슬라이싱
#mpg에서 1~5번째 행과 1~3번째 열을 선택하여 출력하시오.

mpg.iloc[0:5,0:3]
```

```python
#문제 6. loc로 특정 행 + 열 선택
#mpg에서 인덱스가 10인 행의 mpg 값을 출력하시오.

mpg.loc[10,"mpg"]
```
