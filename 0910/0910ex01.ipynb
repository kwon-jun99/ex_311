```python
import pandas as pd

df = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수', '최수진', '정우진'],
    '학년': [1, 2, 1, 3, 2],
    '점수': [85, 92, 78, 95, 88],
    '출석': [90, 95, 80, 98, 85]
})

print(df)
```

```python
df['성적'] = ['B', 'A', 'C', 'A+', 'B+'] # 새로운 열 = 시리즈 타입 / 리스트 -> 시리즈 변경 후 대입

print(df)
display(df)
```

```python
df["생년"]=2000 # 값-> 같은 값의 시리즈로 변경 후 입력

print(df)
```

```python
df['총점'] = df['점수'] + df['출석']

display(df)
```

```python
# 수정
# []: 특정 열 전체 수정
# iloc[], loc[] : 특정위치(헹,열)의 값을 수정
```

```python
df["생년"] =1999

display(df)
```

```python
df.iloc[2,4] = "C+"
df
```

```python
df.loc[2,"성적"] ="C"
df
```

```python
df.loc[[2,4],"출석"] = 88
df
```

```python
df.iloc[[2,4],3] = 85
df
```

```python
df["점수"]>90
```

```python
df.loc[df["점수"]>90,"점수"] = 99
df
```

```python
df = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수', '최수진', '정우진'],
    '학년': [1, 2, 1, 3, 2],
    '점수': [85, 92, 78, 95, 88],
    '출석': [90, 95, 80, 98, 85]
})
```

```python
print(df.describe())
```

```python
df["성적"] = ['B', 'A', 'C', 'A', 'A']  
df["성적"].describe()
```

```python
df['학년'] = df['학년'].astype('category')
df.describe()
df.info()
df["학년"].describe()
```
