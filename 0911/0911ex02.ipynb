```python
import pandas as pd
import matplotlib.pyplot as plt
# 차트에 한글 깨지지 않도록 차트 폰트변경
plt.rcParams['font.family'] = 'Malgun Gothic'
# 차트에 "-"문자 깨짐 방지
plt.rcParams['axes.unicode_minus'] = False
```

```python
df = pd.DataFrame({
    "이름": ["김철수", "이영희", "박민수", "최지영"],
    "성별": ["남", "여", "남", "여"],
    "중간고사": [78, 82, 85, 90],
    "기말고사": [84, 88, 80, 92]
})
df
```

```python
# 전처리 필요 남 -> 평균 / 여 -> 평균
result=df.groupby("성별")[["중간고사","기말고사"]].mean()

result
```

```python
result.plot(kind= "bar")
```

```python
result.plot(kind= "bar", stacked=True) # stacked=False이면 그룹막대 /stacked=True이면 누적막대    
```
