```python
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")
tips.head()
```

```python
fig = plt.figure(figsize=(7,5))
axes1 = fig.add_subplot(1,1,1)

axes1.hist(tips["total_bill"],bins=10)

#글자 삽입

axes1.set_title("Total bill")
axes1.set_xlabel("TB")
axes1.set_ylabel("Count")

#간격 설정
fig.tight_layout()

plt.show()
```

```python
# ============================================================
# 1. 데이터 준비
# ============================================================

# seaborn에서 제공하는 tips 데이터셋을 불러온다.
tips = sns.load_dataset("tips")

# 데이터 확인
print(tips.head())


# ============================================================
# 2. 성별 데이터 분리
# ============================================================

# 성별이 Male인 데이터만 선택한다.
male = tips.loc[tips["sex"] == "Male"]
display(male.head())
# 성별이 Female인 데이터만 선택한다.
female = tips.loc[tips["sex"] == "Female"]
display(female.head())

# ============================================================
# 3. 전체 Figure 생성
# ============================================================

# 전체 그래프를 담을 큰 도화지(Figure)를 생성한다.
fig = plt.figure(figsize=(8, 5))


# ============================================================
# 4. 그래프 공간(Axes) 생성
# ============================================================

# 1행 1열 중 첫 번째 위치에
# 그래프 공간을 생성한다.
axes1 = fig.add_subplot(1, 1, 1)


# ============================================================
# 5. 산점도 그리기
# ============================================================

# 남성 데이터
#
# x축 → total_bill : 전체 식사금액
# y축 → tip        : 팁
#
# label을 지정하면 범례에 표시할 수 있다.
axes1.scatter(
    male["total_bill"],
    male["tip"],
    label="Male"
)


# 여성 데이터
axes1.scatter(
    female["total_bill"],
    female["tip"],
    label="Female"
)


# ============================================================
# 6. 그래프 설정
# ============================================================

# 그래프 제목
axes1.set_title("Total Bill vs Tip by Sex")

# x축 이름
axes1.set_xlabel("Total Bill")

# y축 이름
axes1.set_ylabel("Tip")

# 범례 표시
axes1.legend()


# ============================================================
# 7. Figure 출력
# ============================================================

fig.tight_layout()

plt.show()
```

```python
# ============================================================
# 1. 데이터 준비
# ============================================================

# seaborn에서 제공하는 tips 데이터셋을 불러온다.
tips = sns.load_dataset("tips")

# 데이터 확인
print(tips.head())


# ============================================================
# 2. 전체 Figure 생성
# ============================================================

# 전체 그래프를 담을 큰 도화지(Figure)를 생성한다.
fig = plt.figure(figsize=(8, 5))


# ============================================================
# 3. 그래프 공간(Axes) 생성
# ============================================================

# 1행 1열 중 첫 번째 위치에
# 그래프 공간(Axes)을 생성한다.
axes1 = fig.add_subplot(1, 1, 1)


# ============================================================
# 4. 성별에 따른 tip 박스플롯 그리기
# ============================================================

# 남성의 tip 데이터
male_tip = tips.loc[tips["sex"] == "Male", "tip"]

# 여성의 tip 데이터
female_tip = tips.loc[tips["sex"] == "Female", "tip"]


# 남성과 여성의 tip 분포를 박스플롯으로 비교한다.
axes1.boxplot(
    [male_tip, female_tip],
    
    # 현재 Matplotlib에서는 labels 대신
    # tick_labels를 사용한다.
    tick_labels=["Male", "Female"]
)


# ============================================================
# 5. 그래프 설정
# ============================================================

# 그래프 제목
axes1.set_title("Tip Distribution by Sex")

# x축 이름
axes1.set_xlabel("Sex")

# y축 이름
axes1.set_ylabel("Tip")


# ============================================================
# 6. 그래프 간격 조정
# ============================================================

fig.tight_layout()


# ============================================================
# 7. Figure 출력
# ============================================================

plt.show()
```
