```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
```

```python
arr=np.array([60, 62, 65, 67, 68, 70, 72, 75, 78, 95])
print(arr)
print(type(arr))
```

```python
s= pd.Series(arr)
print(s)
```

```python
Q1= s.quantile(0.25)
Q3= s.quantile(0.75)
median= s.median()
print(Q1,median,Q3)

IQR= Q3-Q1
print("IQR:",IQR)

upper= Q3 + (IQR*1.5)
lower= Q1 - (IQR*1.5)
print("upper:",upper)
print("lower:",lower)
```

```python
fig= plt.figure()
axes1= fig.add_subplot(1,1,1)
sns.boxplot(s)
plt.show()
```

```python
mean = s.mean()
print("mean",mean)
```

```python
var=round(arr.var(),1)
std=round(arr.std(),1)
print("var",var,"std:",std)
```

```python
Z_score=(s-mean) / std
print(Z_score.abs()>3)  #범위 ±3

fig= plt.figure()
axes1=fig.add_subplot(1,1,1)
sns.histplot(s)
plt.show()
```
