```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    "이름": ["홍길동", "김철수"],
    "점수": [80, 90]
})

df2 = pd.DataFrame({
    "이름": ["이영희", "박민수"],
    "점수": [85, 95]
})

display(df1)
display(df2)
```

```python
# 데이터 합치기 concat
df = pd.concat([df1, df2], ignore_index=True, axis=0)
display(df)
```

```python
df = pd.concat([df1, df2],ignore_index=False, axis=1)
display(df)
```

```python
df1 = pd.DataFrame({
    "이름": ["김민수", "이서연"],
    "점수": [80, 90]
})

df2 = pd.DataFrame({
    "이름": ["박지훈", "최유진"],
    "등급": ["A", "B"]
})
display(df1)
display(df2)
```

```python
df=pd.concat([df1,df2])
df
```

```python
df = pd.concat([df1, df2], join="outer", ignore_index=True) # join생략하면 기본값이 outer
display(df)
```

```python
# merge 함수: SQL JOIN 유사

df_customer = pd.DataFrame({
    "고객ID": [101, 102, 103, 104],
    "이름": ["홍길동", "김철수", "이영희", "박민수"]
})

df_score = pd.DataFrame({
    "고객ID": [101, 102, 103, 105],
    "점수": [90, 80, 95, 70]
})


display(df_customer)
display(df_score)

result = pd.merge(
    df_customer,
    df_score,
    on="고객ID",
    how="inner" # inner(기본값), left, right, outer
)

display(result)
```

```python
result = pd.merge(
    df_score,
    df_customer,
    on="고객ID",
    how="left"
)

display(result)
```
