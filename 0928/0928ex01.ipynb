```python
import pandas as pd

data = {
    "날짜": [
        "2025-01-01", "2025-01-02", "2025-01-03",
        "2025-02-01", "2025-02-02", "2025-02-03",
        "2025-03-01", "2025-03-02", "2025-03-03"
    ],
    "매출": [
        100, 120, None,
        200, None, 220,
        300, 320, None
    ]
}

df = pd.DataFrame(data)

df["날짜"] = pd.to_datetime(df["날짜"])

print(df)
print(df.info())
```

```python
df["월"] = df["날짜"].dt.month

df
```

```python
df.groupby("월").mean()
```

```python
cond= (df["월"] ==1)
cond
```

```python
df[cond]["매출"]                                #매출
df.loc[df["월"]==1, "매출"]                     #매출
#mean1= df.loc[df[cond]["매출"]].mean()          #평균
mean1= df.loc[df["월"]==1, "매출"].mean()       #평균
mean1
```

```python
df.loc[df["월"]==2, "매출"]        
mean2= df.loc[df["월"]==2, "매출"].mean()
mean2
df.loc[df["월"]==3, "매출"]        
mean3= df.loc[df["월"]==3, "매출"].mean()
mean3
```

```python
#결측지 전처리

cond=(df["매출"].isna()) & (df["월"]==1)        # 1월 중에 결측지 값 확인
df[cond]
df.loc[cond,"매출"] = mean1                    #1월 결측지 1월 평균값으로 채우기
cond=(df["매출"].isna()) & (df["월"]==2)        # 2월 중에 결측지 값 확인
df[cond]
df.loc[cond,"매출"] = mean2                    #2월 결측지 2월 평균값으로 채우기
cond=(df["매출"].isna()) & (df["월"]==3)        # 3월 중에 결측지 값 확인
df[cond]
df.loc[cond,"매출"] = mean3                    #3월 결측지 3월 평균값으로 채우기
df
```

```python
from sklearn.linear_model import LinearRegression
```

```python
data = {
    "공부시간": [1, 2, 3, 4, 5, 6, 7, 8],
    "게임시간": [5, 4, 4, 3, 3, 2, 2, 1],
    "시험점수": [45, 50, None, 62, 68, None, 80, 88]
}

df = pd.DataFrame(data)

df
```

```python
train = df[df["시험점수"].notna()]
train
```

```python
x_train = train[["공부시간","게임시간"]]
y_train = train["시험점수"]
x_train
y_train
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
print(model.coef_)
print(model.intercept_)
```

```python
df[df["시험점수"].isna()]
```

```python
cond = df["시험점수"].isna()            #시험점수가 NaN 인 것
m=df.loc[cond,["공부시간","게임시간"]]
```

```python
result_y=model.predict(m)
result_y
```

```python
df.loc[cond, "시험점수"] = result_y.round(0)
df
```
