```python
# 판다스 피벗   데이터의 행과 열변경
# 열 -> 행 melt / 행 -> 열 pivot(중복된 값이 있을 경우 pivot_table)
import pandas as pd
```

```python
df = pd.DataFrame({
    "이름": ["철수", "영희"],
    "국어": [80, 90],
    "영어": [70, 95],
    "수학": [90, 85]
})

display(df)
```

```python
df_melt=df.melt(id_vars="이름", var_name="과목", value_name="점수")
df_melt
```

```python
df_pivot= df_melt.pivot(index="이름",columns="과목",values="점수")
df_pivot
```

```python
result=df_pivot.reset_index()      # 인덱스값 초기화(기존 인덱스 컬럼으로 이동)
result
```

```python
#pivot_table

df2 = pd.DataFrame({
    "이름": ["철수", "철수", "철수", "철수", "영희", "영희", "영희"],
    "과목": ["국어", "국어", "영어", "수학", "국어", "영어", "수학"],
    "점수": [80, 50, 70, 90, 90, 95, 85]
})

display(df2)
```

```python
df3=df2.pivot_table(index="이름",columns="과목",values="점수", aggfunc="mean")
df3
```
