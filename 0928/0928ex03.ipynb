```python
import pandas as pd

df = pd.DataFrame({
    "부서": ["영업", "영업", "개발", "개발", "인사"],
    "이름": ["철수", "영희", "민수", "지수", "길동"],
    "매출": [100, 150, 200, 250, 120]
})

display(df)
```

```python
print(df.groupby("부서")["매출"].mean())
print(df.groupby("부서")["매출"].sum())
```

```python
print(df.groupby("부서")["매출"].describe())
```

```python
df.groupby("부서")["매출"].transform("mean")
```

```python
df["부서별평균"] = df.groupby("부서")["매출"].transform("mean")
df
```

```python
df2 = pd.DataFrame({
    "부서": ["영업", "영업", "개발", "개발", "인사"],
    "이름": ["철수", "영희", "민수", "지수", "길동"],
    "매출": [100, 150, 200, 250, 120]
})

df2
```

```python
df_group=df2.groupby("부서")["매출"].mean()
df_group
```

```python
df_group2=df_group.reset_index()
df_group3=df_group2.rename(columns={"매출":"평균"})
df_group3
```

```python
df2.merge(df_group3,on="부서")
```
