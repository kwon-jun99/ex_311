```python
import pandas as pd

score = pd.Series(
    [78, 85, 82, 90, 88, 92],
    index=["1학년중간", "1학년기말",
           "2학년중간", "2학년기말",
           "3학년중간", "3학년기말"]
)

print(score)
```

```python
# 문제 1. Series 기본 정보

score.info()
```

```python
# 문제 2. 학년별 성적 조회


score.loc["1학년중간"]
score.loc["2학년기말"]
score.loc["3학년중간"]
```

```python
# 문제 3. iloc를 이용한 조회

score.iloc[0]
score.iloc[-1]
score.iloc[0:2]
score.iloc[4:6]
```

```python
# 문제 4. 성적 슬라이싱

score.iloc[:3]
score.iloc[-3:]
score.iloc[1::2]
score.iloc[::-1]
```
